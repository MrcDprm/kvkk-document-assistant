# KVKK Document Assistant: Turkish NER training

Fine-tunes two small Turkish BERT models (mini and small) on WikiANN (Turkish) and synthetic document/question sentences, then exports each as an int8 ONNX model.

**Before running:** Runtime → Change runtime type → **T4 GPU**. Then Runtime → Run all.

At the end the browser downloads `kvkk-ner-mini.zip` and `kvkk-ner-small.zip`.

```python
import json
for name in ('mini', 'small'):
    m = json.load(open(f'/content/kvkk-ner-{name}/metrics.json'))
    for stage in ('pytorch', 'onnx_int8'):
        for dataset, report in m[stage].items():
            print(f"{name:6} {stage:10} {dataset:10} PER F1 {report['PER']['f1']:.3f}  ALL F1 {report['ALL']['f1']:.3f}")
    print(f"{name:6} parameters {m['parameters']:,}  training {m['training_seconds']} s")
```
